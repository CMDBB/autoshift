```python
import pulp
```

```python
prob = pulp.LpProblem("test", pulp.LpMaximize)

x1 = prob.add_variable("x1", 0, None)

x2 = prob.add_variable("x2", 0, None)

o = prob.add_variable("o", 0, None)
u = prob.add_variable("u", 0, None)
prob += x1 - x2 == o - u

prob += 3 * x1 + 2 * x2 <= 10
prob += 2 * x1 + 3 * x2 <= 10

prob += 6 * x1 + 1 * x2 - 2 * (o + u)
# prob += 6*x1 + 5.8*x2

prob.solve()
for v in prob.variables():
	print(f"{v!s:>3}|{pulp.value(v):>5.3}|{v.dj:>5.3}")
for c in prob.constraints():
	print(f"{c!s:>30}|{c.pi:>5.3}")
```
